#### 3.1.2: Implement a Linear Regression Model

````python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import MaxAbsScaler

#Encode
df_3 = pd.read_csv('diabetic_data.csv') #Loading in dataset again as to contain everything to this code block
df_3.replace("?", np.nan, inplace=True)

df_3.drop(columns=['weight', 'payer_code', 'medical_specialty', 'max_glu_serum', 'A1Cresult'], inplace=True)
df_3.dropna(axis=0, how='any', inplace=True)

LE = LabelEncoder()
OE = OneHotEncoder(sparse_output=False)

columns_LE = ["age", "race", "gender", 'metformin','repaglinide','nateglinide','chlorpropamide','glimepiride','acetohexamide','glipizide','glyburide','tolbutamide','pioglitazone','rosiglitazone','acarbose','miglitol','troglitazone','tolazamide','examide','citoglipton','insulin','glyburide-metformin','glipizide-metformin','glimepiride-pioglitazone','metformin-rosiglitazone','metformin-pioglitazone']
for column in columns_LE:
    df_3[column] = LE.fit_transform(df_3[column])

columns_OE= ["change", 'diag_1','diag_2','diag_3', "diabetesMed", "readmitted"]
for column in columns_OE:
    df_3[column] = OE.fit_transform(df_3[[column]])

#feature scaling
MMS = MinMaxScaler()
df_3[df_3.columns] = MMS.fit_transform(df_3[df_3.columns])

X_3 = df_3.drop(columns=['age'])
y_3 = df_3['age']

X_train_3, X_test_3, y_train_3, y_test_3 = train_test_split(X_3, y_3, test_size=0.2, random_state=42)

LR_3 = LinearRegression()
LR_3.fit(X_train_3, y_train_3)
LR_pred_3 = LR_3.predict(X_test_3)

#add MAE and MSE
print(mean_absolute_error(y_test_3, LR_pred_3))
print(mean_squared_error(y_test_3, LR_pred_3))
````

### **Task 3.3: Implementing a Clustering Model**


#### 3.3.1: Load and Preprocess the Data (Already done in previous tasks)

````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import MaxAbsScaler
from sklearn.preprocessing import StandardScaler

#Encode
df_3 = pd.read_csv('diabetic_data.csv') #Loading in dataset again as to contain everything to this code block
df_3.replace("?", np.nan, inplace=True)

df_3.drop(columns=['weight', 'payer_code', 'medical_specialty', 'max_glu_serum', 'A1Cresult'], inplace=True)
df_3.dropna(axis=0, how='any', inplace=True)

LE = LabelEncoder()
OE = OneHotEncoder(sparse_output=False)

columns_LE = ["age", "race", "gender", 'metformin','repaglinide','nateglinide','chlorpropamide','glimepiride','acetohexamide','glipizide','glyburide','tolbutamide','pioglitazone','rosiglitazone','acarbose','miglitol','troglitazone','tolazamide','examide','citoglipton','insulin','glyburide-metformin','glipizide-metformin','glimepiride-pioglitazone','metformin-rosiglitazone','metformin-pioglitazone']
for column in columns_LE:
    df_3[column] = LE.fit_transform(df_3[column])

columns_OE= ["change", 'diag_1','diag_2','diag_3', "diabetesMed", "readmitted"]
for column in columns_OE:
    df_3[column] = OE.fit_transform(df_3[[column]])

#feature scaling
MMS = MinMaxScaler()
df_3[df_3.columns] = MMS.fit_transform(df_3[df_3.columns])

scaler_3 = StandardScaler()
df_3_scaled = scaler_3.fit_transform(df_3)
````

#### 3.3.2: Implement K-Means Clustering

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertia_values = []
cluster_range = range(2, 11)

for k in cluster_range:
    KM_3 = KMeans(n_clusters=k, random_state=42)
    KM_3.fit(df_3)
    inertia_values.append(KM_3.inertia_)
    labels = KM_3.labels_
    sil_score = silhouette_score(df_3, labels)
    print(f"Silhouette Score for {k} clusters: {sil_score}")


plt.figure(figsize=(8, 5))
plt.plot(cluster_range, inertia_values, marker='o', linestyle='--')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal Clusters')
plt.xticks(cluster_range)
plt.grid()
plt.show()
#5 clusters seems to be the optimal solution after running above code.

KM_3 = KMeans(n_clusters=5, init='k-means++', n_init='auto', max_iter=300, tol=0.0001, verbose=0, random_state=42, copy_x=True, algorithm='lloyd')
KM_3.fit(df_3)
cluster_labels = KM_3.labels_
print(silhouette_score(df_3, cluster_labels))
````
